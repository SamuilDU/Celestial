```python
from math import comb, factorial

def LegendrePolynomialCoefficients(l, m):
    
    polynomial_coeffcients, exponent = np.zeros(l), np.zeros(l)
    
    if np.abs(m) > l:
        raise ValueError("The value of m must be between 0 and l")

    if l < 0:
        raise ValueError("l must be a positive integer")
    

    if m < 0:
        m *= -1

        for k in range(l + 1):
           exponent[l  -1 - k] = 2*l - m - k

           if exponent[l - 1 - k] < 0:
               break    
              
           polynomial_coeffcients[k - 1] = (-1)**(l + m) * comb(l, k) * aux_cumprod(2*l - m, 0, l + m + 1, 1)
    
        polynomial_coeffcients *= (factorial(l) * 2**l)**(-1) * (-1)**m *factorial(l - m)/factorial(l + m) 

    return polynomial_coeffcients, exponent

def LegendrePolynomialValues(x, l, m):
    coefficients, exponents = LegendrePolynomialCoefficients(l, m)

    for val in x:
        x = (1 - val*val)**(np.abs(m)/2)* sum(coefficients * val **(exponents))

    return x

LegendrePolynomialCoefficients(1, -1)
```
